```python
# Calculate pi with monte carlo
import numpy as np

N = 10**7

X = 2.0*(np.random.rand(2,N)-0.5)
r = np.sqrt(np.power(X[0,:],2)+np.power(X[1,:],2))

Y = np.zeros(N)

Y[r < 1.0] = 4.0
print('%.3f'%np.mean(Y))
```

```python
# monte carlo hello world

import numpy as np
N = 10**6

X = np.random.rand(N) # generate random number between 0 to 1
Y = 5.0 * np.power(X,4.0)

print("%.3f" %np.mean(Y)) # sum all the values of Y and find the mean value (divided by count)
```
