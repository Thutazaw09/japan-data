````python
# to query the realsense device and return as a list

print("List devices")
print("----------------")

import pyrealsense2 as rs

# Create a context object
ctx = rs.context()
devices = ctx.devices

if len(devices) == 0:
    print("❌ No RealSense devices detected.")
else:
    print("✅ Found", len(devices), "device(s):\n")
    for d in devices:
        print("Device Name      :", d.get_info(rs.camera_info.name))
        print("Serial Number    :", d.get_info(rs.camera_info.serial_number))
        print("Firmware Version :", d.get_info(rs.camera_info.firmware_version))
        print("USB Type         :", d.get_info(rs.camera_info.usb_type_descriptor))
        print("-" * 40)

````

#### Note

`context()`
> *This is a top level object that represent the current connection to all available RealSense devices and provides access to the SDK's global state.</br> *It queries and manages all connected devices. </br> *It aslo allows to monitor device events. 
</br> *Key functions on this object are 
</br> `query_devices()` returns a device_list of all currently connected camera.
</br>`devices` short hand property for return a device_list of all currently connected camera.
</br>`set_devices_changed_callback(callback)` lets you register a function that triggers if deviecs are plugged/unplugged.
</br>`load_device(file_path)` load a ```.bag``` file recording file as if it were a live device, lets run a same pipeline code on recorded streams.
</br>`ctx.get_api_version()` return SDK API version.
</br>`camera_info` is the object that store the camera information.
